# 📊 Ad Campaign Insights Dashboard

This notebook loads ad campaign data, computes key metrics (CTR, CPC, CPA), and displays insights for optimization.

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("ad_campaign_data.csv")

# Preview data
df.head()
```

```python
# Calculate performance metrics
df['CTR'] = (df['Clicks'] / df['Impressions']) * 100
df['CPC'] = df['Cost'] / df['Clicks']
df['CPA'] = df['Cost'] / df['Conversions']

# Round values
df[['CTR', 'CPC', 'CPA']] = df[['CTR', 'CPC', 'CPA']].round(2)

# Preview updated DataFrame
df.head()
```

```python
# Group by Campaign and summarize
summary = df.groupby('Campaign_Name').agg({
    'Impressions': 'sum',
    'Clicks': 'sum',
    'Conversions': 'sum',
    'Cost': 'sum',
    'CTR': 'mean',
    'CPC': 'mean',
    'CPA': 'mean'
}).round(2).sort_values(by='Conversions', ascending=False)

summary
```

```python
# Basic insights
top_ctr = summary.sort_values(by='CTR', ascending=False).head(3)
lowest_roi = summary.sort_values(by='CPA', ascending=False).head(3)

print("🔼 Top 3 Campaigns by CTR:")
display(top_ctr)

print("\n🔻 Bottom 3 Campaigns by ROI (High CPA):")
display(lowest_roi)
```
