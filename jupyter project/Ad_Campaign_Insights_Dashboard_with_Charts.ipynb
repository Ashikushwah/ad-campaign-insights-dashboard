# 📊 Ad Campaign Insights Dashboard

This notebook loads ad campaign data, computes key metrics (CTR, CPC, CPA), displays insights, and visualizes performance metrics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV file
df = pd.read_csv("ad_campaign_data.csv")

# Preview data
df.head()
```

```python
# Calculate performance metrics
df['CTR'] = (df['Clicks'] / df['Impressions']) * 100
df['CPC'] = df['Cost'] / df['Clicks']
df['CPA'] = df['Cost'] / df['Conversions']

# Round values
df[['CTR', 'CPC', 'CPA']] = df[['CTR', 'CPC', 'CPA']].round(2)

# Preview updated DataFrame
df.head()
```

```python
# Group by Campaign and summarize
summary = df.groupby('Campaign_Name').agg({
    'Impressions': 'sum',
    'Clicks': 'sum',
    'Conversions': 'sum',
    'Cost': 'sum',
    'CTR': 'mean',
    'CPC': 'mean',
    'CPA': 'mean'
}).round(2).sort_values(by='Conversions', ascending=False)

summary
```

```python
# Basic insights
top_ctr = summary.sort_values(by='CTR', ascending=False).head(3)
lowest_roi = summary.sort_values(by='CPA', ascending=False).head(3)

print("🔼 Top 3 Campaigns by CTR:")
display(top_ctr)

print("\n🔻 Bottom 3 Campaigns by ROI (High CPA):")
display(lowest_roi)
```

```python
# Visualization: CTR by Campaign
plt.figure(figsize=(10, 6))
summary['CTR'].sort_values().plot(kind='barh', color='skyblue')
plt.title('Average CTR by Campaign')
plt.xlabel('CTR (%)')
plt.ylabel('Campaign Name')
plt.tight_layout()
plt.show()
```

```python
# Visualization: Conversions by Campaign
plt.figure(figsize=(10, 6))
summary['Conversions'].sort_values().plot(kind='barh', color='green')
plt.title('Total Conversions by Campaign')
plt.xlabel('Conversions')
plt.ylabel('Campaign Name')
plt.tight_layout()
plt.show()
```
